```python
from pypdf import PdfReader
import os

documents = []

for pdf_path in pdf_files:
    reader = PdfReader(pdf_path)

    text = ""

    for page in reader.pages:
        text += page.extract_text() or ""

    documents.append({
        "source": os.path.basename(pdf_path),
        "content": text
    })

print("Total documents loaded:", len(documents))

for doc in documents:
    print("\nDocument:", doc["source"])
    print("Characters:", len(doc["content"]))
```

```python
chunks = []

chunk_size = 800
overlap = 150

for doc in documents:
    text = doc["content"]
    source = doc["source"]

    start = 0

    while start < len(text):
        end = start + chunk_size

        chunk_text = text[start:end]

        if chunk_text.strip():
            chunks.append({
                "content": chunk_text,
                "source": source
            })

        start += chunk_size - overlap

print("Total chunks created:", len(chunks))

print("\nFirst chunk:")
print(chunks[0]["content"])

print("\nSource:", chunks[0]["source"])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

chunk_texts = [chunk["content"] for chunk in chunks]
vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(chunk_texts)

print("Search system created successfully!")
print("Total searchable chunks:", tfidf_matrix.shape[0])
print("Number of text features:", tfidf_matrix.shape[1])
```

```python
def retrieve_documents(question, top_k=3):

    
    question_vector = vectorizer.transform([question])

    
    scores = cosine_similarity(question_vector, tfidf_matrix).flatten()

    
    top_indices = scores.argsort()[-top_k:][::-1]

    results = []

    for index in top_indices:
        results.append({
            "content": chunks[index]["content"],
            "source": chunks[index]["source"],
            "score": scores[index]
        })

    return results


print("HR document retrieval function created!")
```

```python
import pandas as pd
import time

answers = []

for index, row in test_df.iterrows():

    question_id = row["question_id"]
    question = row["question"]

    print(f"\nProcessing {question_id}: {question}")

    
    results = retrieve_documents(question, top_k=5)


    context = "\n\n".join([
        f"Source: {doc['source']}\n{doc['content']}"
        for doc in results
    ])

    prompt = f"""
You are the HR Help Desk Assistant for Zyro Dynamics.

Answer employee questions using ONLY the provided HR policy context.

Rules:
1. Never invent information, numbers, dates or policies.
2. If the context does not contain the answer, respond:
   "Sorry, I could not find this information in the available HR policies."
3. For questions unrelated to Zyro Dynamics HR policies, politely refuse.
4. Give clear, concise and accurate answers.
5. Do not use outside knowledge.

HR POLICY CONTEXT:
{context}

EMPLOYEE QUESTION:
{question}

ANSWER:
"""

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=prompt
        )

        answer = response.text.strip()

    except Exception as e:
        print("ERROR:", str(e))
        answer = "Sorry, I could not find this information in the available HR policies."

    answers.append({
        "question_id": question_id,
        "answer": answer
    })

    print("ANSWER:", answer)

    time.sleep(1)


answers_df = pd.DataFrame(answers)


submission = sample_df[["question_id"]].merge(
    answers_df,
    on="question_id",
    how="left"
)


submission.to_csv("/kaggle/working/submission.csv", index=False)

print("\nSUBMISSION CREATED SUCCESSFULLY!")
display(submission)

print("Shape:", submission.shape)
print("Columns:", submission.columns.tolist())
```

```python
import re
import pandas as pd
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

draft_answers = []

for _, row in test_df.iterrows():
    question = row["question"]
    qid = row["question_id"]

    results = retrieve_documents(question, top_k=5)

    sentences = []

    for doc in results:
        text = doc["content"]
        parts = re.split(r'(?<=[.!?])\s+', text)

        for sentence in parts:
            words = set(re.findall(r'\b[a-zA-Z]{3,}\b', sentence.lower()))
            keywords = set(re.findall(r'\b[a-zA-Z]{3,}\b', question.lower()))
            keywords -= ENGLISH_STOP_WORDS

            score = len(words & keywords)

            if score > 0:
                sentences.append((score, sentence.strip()))

    sentences.sort(key=lambda x: x[0], reverse=True)

    answer = " ".join(
        sentence for _, sentence in sentences[:4]
    )

    if not answer:
        answer = "This information is not available in the retrieved HR policy text."

    draft_answers.append({
        "question_id": qid,
        "question": question,
        "answer": answer
    })

draft_df = pd.DataFrame(draft_answers)

display(draft_df)
```

```python
print("Total answers:", len(draft_df))

print("\nEmpty answers:")
print(draft_df[draft_df["answer"].str.strip() == ""])

print("\nUnavailable answers:")
display(
    draft_df[
        draft_df["answer"].str.contains(
            "not available",
            case=False,
            na=False
        )
    ][["question_id", "question", "answer"]]
)

print("\nDuplicate question IDs:",
      draft_df["question_id"].duplicated().sum())
```
